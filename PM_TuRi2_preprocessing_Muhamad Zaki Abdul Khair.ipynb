# Latihan Mandiri

```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

# Latihan Mandiri 1

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
print("Mean usia   :", df['usia'].mean())
print("Median usia :", df['usia'].median())

```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']

```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2})
X = pd.get_dummies(X, columns=['kota'], dtype=int)
y = y.map({'Tidak':0,'Ya':1})
```

# Latihan Mandiri 2

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler
num = ['usia','pendapatan','pendidikan']
sc = MinMaxScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
X_train.describe()
```

# Refleksi


1. karena bagi data menjadi latih & uji SEBELUM penskalaan. Scaler di fit hanya pada data latih, lalu transform pada keduanya.
jika scaler di-fit pada seluruh data sebellum di split maka akan ada kebocoran informasi ke pelatihan.
2. ada encoding Label kalo datanya bentuk ordinal (berurtan; contoh pendidikan: SD, SMP, SMA. harus berurutan) hasilnya satu kolom angka (SMA beri nilai 1, S1 beri nilai 2, S2 nilainya 3. Resikonya urutan palsu contohnya urutannya tidak seperti itu misal tdi nilai 1,2,3 malah 3,2,1)
- one-hot encoding kalo datanya bentuk nominal (tanpa urutan; contoh: Bandung/Jakarta/Surabaya) hasilnya banyak kolom biner (0/1).Resikonya kolom membengkak, contoh di iris berupa target berupa jenis bunga menjadi 3, kalo label nya 100 maka jadi 100.

3. teknik normalisasi (minmax)
- skala ke rentang tetep 0-1. misal data 100 menjadi 0, data 1000 menjadi 1 karena maximumnya
- berguna bila butuh batas jelas.
- sensitif terhadap oulier
teknik standardisasi (Z-score)
- ubah ke mean 0, std 1.
- tidak membatasi rentang
- lebih tahan terhadap outlier --> pilihan default
